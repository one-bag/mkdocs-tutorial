# Example notebook

This is a real Jupyter notebook, stored in the `docs` folder as
`example-notebook.ipynb`. MkDocs turns it into the page you are reading,
with the code, the printed output and the chart all in place.

The numbers below are made up. What matters is that everything you see
under a code cell was produced by running that cell.

## A small table

Markdown cells become ordinary page text. Code cells keep their output.

```python
import pandas as pd

pages = pd.DataFrame(
    {
        "month": ["January", "February", "March", "April", "May", "June"],
        "pages_published": [3, 5, 4, 8, 6, 9],
        "edits": [12, 21, 17, 34, 25, 41],
    }
)

pages
```

```python
print(f"Total pages published: {pages['pages_published'].sum()}")
print(f"Busiest month:         {pages.loc[pages['edits'].idxmax(), 'month']}")
```

## A chart

Images produced by a cell are stored inside the notebook file, so they
appear on the site without any extra work.

The figure gets an explicit white background. Without it the chart would
be transparent, and its dark axis labels would be unreadable when a
visitor switches the site to dark mode.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 3.5), facecolor="white")
ax.bar(pages["month"], pages["pages_published"], color="#00695c")
ax.set_ylabel("Pages published")
ax.set_title("Example data, not real measurements")
ax.spines[["top", "right"]].set_visible(False)
ax.set_facecolor("white")
fig.tight_layout()
plt.show()
```

## What to notice

- The first heading of the notebook became the title of this page.
- The table and the chart are the outputs saved in the notebook file.
  Nothing was executed while the site was being built.
- Re-run the notebook, save it, commit it, and this page changes.
